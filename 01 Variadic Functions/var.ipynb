# Variadic Functions

```python
def add(a: int | float, b: int | float) -> int | float:
    """ "
    This function adds two numbers
    """
    c = a + b
    return c
```

```python
# positional arguments
a1 = add(3, 4)
```

```python
print(a1)
```

```python
# Keyword arguments
a2 = add(a=23.5, b=12.7)
```

```python
print(a2)
```

# Variadic functions

```python
# *args positional arguments -> acts like a tuple in any function
```

```python
def func1(*args):
    print(args)
    print(f"Type : {type(args)}")
    for i in args:
        print(i)
```

```python
func1(2, 3)
```

```python
func1(12)
```

```python
func1(6, 7)
```

```python
func1(4, 5, 6)
```

```python
def add_multiple(*args):
    # intialize sum to 0
    s = 0
    # apply for loop on args
    for i in args:
        s = s + i
    # return the values of s
    return s
```

```python
add_multiple(9, 3)
```

```python
add_multiple(4, 5, 67, 45)
```

```python
add_multiple(11, 12, 13, 14, 100, 1000, 2001)
```

```python
a = [12, 34, 56, 78, 90, 23, 45, 67, 89, 1, 10, 100, 1000, 10000]
```

```python
add_multiple(*a)
```

```python
b = list(range(1, 1000))
```

```python
print(b)
```

```python
add_multiple(*b)
```

```python
c = list(range(1, 10000))
```

```python
print(c)
```

```python
add_multiple(*c)
```

## Keyword arguments
**kwargs -> keyword arguments -> They act like a dictionary in python

```python
def func2(**kwargs):
    print(kwargs)
    print(f"Type: {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"Key : {key}, Value : {value}")
```

```python
func2(a=1, b=23.4, name="Raman", test=False)
```

```python
func2(name="Akshay")
```

```python
func2(name="Rahul", age=23, email="a@gmail.com", married=True)
```

## You can also unpack dictionary as kwargs with **

```python
d = {"a": 23, "b": 45, "c": "Example"}
```

```python
type(d)
```

```python
func2(**d)
```

```python
def create_user(**kwargs):
    user = {"Name": None, "Age": None, "City": None}
    user.update(kwargs)
    return user
```

```python
u1 = create_user(name="Sarthak", age=23, city="Pune")
print(u1)
```

```python
u2 = create_user(Name="Rahul", Age=34)
```

```python
print(u2)
```

```python
u3 = create_user(Name="Aditi", Age=23, Contact=987654321, Blood_grp="A+ve")
print(u3)
```

```python
u4 = create_user(name="Raman", age=35, city="Banglore", email="raman@test.com")
print(u4)
```

## Use *args and **kwargs both

Postional arguments are always first inputs, then you have kwargs


```python
def func3(*args, **kwargs):
    # args
    print(args)
    print(f"Type of args : {type (args)}")
    for i in args:
        print(i)
    # kwargs
    print(kwargs)
    print(f"Type of kwargs : {type(kwargs)}")
    for key, value in kwargs.items():
        print(f"Key : {key}, Value : {value}")
```

```python
func3(1, 2, 3, 4, 6)
```

```python
func3(name="Aditi", age=23, city="Pune")
```

```python
# incorrect because postional arguments should be first
func3(a=23, b=24, 13, 14, 15)
```

```python
func3(a=12, b=13, c=14, d=15)
```

```python
func3(12, 13, 14, 15, a=12, b=13, c=14, d=15)
```

## Variadic functions provide flexibity in input to user
1) *args -> Postional arguments act like a tuple
2) **kwargs -> Keyword arguments act like a dictionary
