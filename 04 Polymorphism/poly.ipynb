# Assignment - Create a class triangle to calculate area / perimeter

```python
from dataclasses import dataclass
import math


@dataclass
class Triangle:
    a: int | float
    b: int | float
    c: int | float

    def __post_init__(self):
        if (self.a <= 0) or (self.b <= 0) or (self.c <= 0):
            raise ValueError("Sides of triangle cannot be negative or 0")
        if not self.is_triangle():
            raise ValueError("Sum of two sides should be greater than third")

    def is_triangle(self) -> bool:
        return (
            (self.a + self.b > self.c)
            and (self.b + self.c > self.a)
            and (self.a + self.c > self.b)
        )

    def perimeter(self) -> int | float:
        return self.a + self.b + self.c

    def area(self) -> float:
        s = self.perimeter() / 2
        a = math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))
        return a
```

## Polymorphism in classes
different classes but same functions

```python
class India:
    def capital(self):
        print("New Delhi is capital of India")

    def language(self):
        print("Hindi is widly spoken in India")
```

```python
class USA:
    def capital(self):
        print("Washington D. C. is capital of USA")

    def language(self):
        print("English is widely spoken in USA")
```

```python
class France:
    def capital(self):
        print("Paris D. C. is capital of France")

    def language(self):
        print("French is widely spoken in France")
```

```python
c1 = India()
```

# Implenting polymorphism for different shapes

```python
@dataclass
class Rectangle:
    width: int | float
    height: int | float

    def perimeter(self) -> int | float:
        return 2 * (self.width + self.height)

    def area(self) -> int | float:
        return self.width * self.height
```

```python
import math
```

```python
math.pi
```

```python
@dataclass
class Circle:
    radius: int | float

    def perimeter(self) -> float:
        return 2 * math.pi * self.radius

    def area(self) -> float:
        return math.pi * (self.radius**2)
```

## Polymorphism example from data analyst point of view

```python
import numpy as np


class NumericalSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        return {
            "mean": np.mean(self.data),
            "max": np.max(self.data),
            "min": np.min(self.data),
            "std": np.std(self.data),
        }
```

```python
class CatgoricalSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        unique_vals, count = np.unique(self.data, return_counts=True)
        return dict(zip(unique_vals, count))
```

```python
# Numerical data
marks = [85, 93, 86, 72, 63, 78, 95]
a = NumericalSummary(data=marks)
print(a.summary())
```

```python
temp = (12.3, 11.5, 17.1, 23, 25)
b = NumericalSummary(data=temp)
print(b.summary())
```

```python
grades = ["A", "B", "A", "B", "C", "C", "B", "F", "A", "B"]
c = CatgoricalSummary(data=grades)
print(c.summary())
```
